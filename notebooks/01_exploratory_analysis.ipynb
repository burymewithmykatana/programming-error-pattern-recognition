# Exploratory Analysis

This notebook inspects the generated Python error-pattern dataset used by the baseline experiments.

```python
from pathlib import Path

import pandas as pd

dataset_path = Path('../data/raw/curated_python_errors.csv')
frame = pd.read_csv(dataset_path)
frame.head()
```

## Label Balance

The generated dataset is designed to be balanced across the eight supported classes.

```python
label_counts = frame['label'].value_counts().sort_index()
label_counts
```

```python
label_counts.plot(kind='bar', title='Examples per Label', ylabel='count')
```

## Example Snippets

Reviewing representative snippets helps verify that labels are understandable and distinct enough for baseline experimentation.

```python
for label, group in frame.groupby('label'):
    print(f'\n--- {label} ---')
    print(group.iloc[0]['code'])
```
